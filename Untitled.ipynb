```python
import pandas as pd
import plotly.express as px

data = pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/d51iMGfp_t0QpO30Lym-dw/automobile-sales.csv')
recession_data = data[data['Recession'] == 1]
recession_data
```

```python
yearly_rec=recession_data.groupby('Year')['Automobile_Sales'].mean().reset_index()
yearly_rec
```

```python
average_sales = recession_data.groupby('Vehicle_Type')['Advertising_Expenditure'].sum().reset_index()
average_sales
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import plotly.express as px

total_sales = recession_data.groupby('Vehicle_Type')['Advertising_Expenditure'].sum().reset_index()
fig = px.pie(data, values='Advertising_Expenditure', names='Vehicle_Type')
fig.show()
```

```python
import plotly.express as px

# Loading the iris dataset
df = px.data.iris()

fig = px.pie(df, values="sepal_width",
             names="species", 
             color_discrete_sequence=px.colors.sequential.RdBu)
fig.show()
```

```python
unemployment_sales = recession_data.groupby(['unemployment_rate', 'Vehicle_Type'])['Automobile_Sales'].sum().reset_index()
unemployment_sales
```

```python
fig = px.bar(unemployment_sales, 
             x='Vehicle_Type', 
             y='Automobile_Sales',
             labels={'unemployment_rate': 'Unemployment Rate', 'Automobile_Sales': 'Average Automobile Sales'},
             title='Effect of Unemployment Rate on Vehicle Type and Sales',
             color='Vehicle_Type')
fig.show()
```
